SQLite database 

```python
from sqlalchemy import create_engine
import time
engine = create_engine('sqlite:///../data/olist_ecommerce.db')
print("database engine created successfully")
```

```python
data_path='../data/'
start_time=time.time()

for file in os.listdir(data_path):
    if file.endswith('.csv'):
        table_name=file.replace('.csv','')
        df=pd.read_csv(data_path + file)
        df.to_sql(table_name,con=engine,if_exists='replace',index=False)
        print(f"loaded:{table_name}-->{df.shape[0]:,} rows")
end_time=time.time()
print(f"\n all tables loaded ! Time taken :{round((end_time-start_time)/60,2)} minutes")
```

```python
import sqlite3

conn = sqlite3.connect('../data/olist_ecommerce.db')
tables = pd.read_sql("SELECT name FROM sqlite_master WHERE type='table'", conn)
print("Tables in database:")
print(tables)
```
